```python
def initialize_inventory(products: list):
    '''Initialize the inventory'''
    inventory = {}
    for prod in products:
        prod_quant= input(f"Quantity of product {prod}")
        inventory[prod] = int(prod_quant)
    return inventory
```

```python

# products = ["A", "B", "C", "D", "E"]
from pydantic_core import ErrorTypeInfo


products_input = input("Enter list of products separated by space")
try:
    products = products_input.split()
except:
    raise ErrorTypeInfo("Invalid Input")

begin_inventory = initialize_inventory(products)
begin_inventory
```

```python
def get_customer_orders():
    '''Enter Product Name'''
    customer_orders = set()
    cust_prod = input(f"Enter first product: ")
    add_another_prod = True
    # more_prod = ["y","yes", "true",1, "Y", "Yes","True"]

    while add_another_prod == True:
        available_prod = False
        while available_prod == False :                   
            if cust_prod in products:
                customer_orders.add(cust_prod)
                available_prod = True
            else:
                print(f"The product {cust_prod} is not available")
                cust_prod = input("Next Product: ") 

        add_prod_input = input("Do you want to add another product? (yes/no)")
        if add_prod_input.lower() == "yes" or add_prod_input.lower() == "true" or add_prod_input.lower() == "y" or add_prod_input == "1":
        #   if add_another_prod in more_prod:  
            cust_prod = input("Next Product: ")
        else:
            add_another_prod = False

    return customer_orders
```

```python
customer_orders = get_customer_orders()
# customer_orders
```

```python
def update_inventory(customer_orders: set, inventory: list):
    '''Update inventory dict based on customer orders'''
    new_inventory = {}
    for key, val in inventory.items():
        if key in customer_orders:    
            new_inventory[key] = val -1
        elif inventory[key] == 0:
            new_inventory[key] = 0
        else:
            new_inventory[key] = val
            
    return new_inventory
```

```python
new_inventory = update_inventory(customer_orders=customer_orders, inventory=begin_inventory)
# new_inventory
```

```python
def calculate_order_statistics(customer_orders: set, products: list):
    '''calculate  order statistics (total products ordered and percentage of unique products ordered'''
    total_prod_ordered = len(customer_orders)
    total_prods = len(set(products))
    percent_unique_order = {}
    for order in customer_orders:
        percent_unique_order[order]= round(100/total_prods,2)
    return [total_prod_ordered, percent_unique_order]
```

```python
statistics = calculate_order_statistics (customer_orders=customer_orders, products = products )
```

```python
def print_order_statistics(order_statistics):
    print(f"Total prod ordered: {order_statistics[0]}")
    print(f"Percentage of unique products ordered {order_statistics[1]}")
```

```python
print_order_statistics(order_statistics = statistics)
```

```python
def print_updated_inventory(inventory: dict):
    print(f"Updated inventory: {inventory}")
```
